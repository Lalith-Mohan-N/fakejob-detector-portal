```python
# SHAP explanation demo
import joblib
import shap
from app.ml_models.preprocess import EMSCADPreprocessor
from app.core.config import settings

preprocessor = EMSCADPreprocessor.load(settings.MODEL_DIR / 'preprocessor')
rf = joblib.load(settings.MODEL_DIR / 'random_forest.joblib')

# Sample a few real rows for background data
import pandas as pd
df = pd.read_csv('../data/raw/fake_job_postings.csv').sample(100, random_state=42)
X, _ = preprocessor.transform(df)

explainer = shap.TreeExplainer(rf)
shap_values = explainer.shap_values(X)
shap.summary_plot(shap_values, X, feature_names=preprocessor.tfidf.get_feature_names_out().tolist()[:20])
```
