# EMSCAD Dataset EDA

Exploratory data analysis on the real Employment Scam Aegean Dataset (EMSCAD).

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

CSV_PATH = Path('../data/raw/fake_job_postings.csv')
df = pd.read_csv(CSV_PATH)
df.head()
```

```python
print('Shape:', df.shape)
print('Columns:', df.columns.tolist())
df.info()
```

```python
# Class distribution
sns.countplot(x='fraudulent', data=df)
plt.title('Real vs Fraudulent Job Postings')
plt.show()
print(df['fraudulent'].value_counts(normalize=True))
```

```python
# Text length distribution by class
df['description_length'] = df['description'].fillna('').apply(len)
sns.boxplot(x='fraudulent', y='description_length', data=df)
plt.title('Description Length: Real vs Fraudulent')
plt.show()
```

```python
# Missing values heatmap
sns.heatmap(df.isnull(), cbar=False, yticklabels=False)
plt.title('Missing Values in EMSCAD')
plt.show()
```
